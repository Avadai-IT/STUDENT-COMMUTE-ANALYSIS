```python
import pandas as pd

url = "https://raw.githubusercontent.com/theTanmayAgarwal/Student-Commute-Analysis/main/student_commute_data_google_form.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.info()


df.isnull().sum()
df.isnull().sum().sum()
df["Major Delay Points / Bottlenecks on this Route"].isnull().sum()
print("Average travel time:",
      df["Average Total Travel Time (in minutes)"].mean())

print("Average one-way cost:",
      df["Average Total One-Way Cost (in ₹)"].mean())

print("Average satisfaction:",
      df["Overall Satisfaction with this Commute"].mean())
print(
    df.groupby("Primary Mode of Transport")
      ["Average Total Travel Time (in minutes)"]
      .mean()
      .sort_values()
)
df.groupby("Primary Mode of Transport")[
    "Average Total One-Way Cost (in ₹)"
].mean().sort_values()
df.groupby("Time of Day You Usually Travel")[
    "Average Total Travel Time (in minutes)"
].mean().sort_values()
df["Major Delay Points / Bottlenecks on this Route"].value_counts().head(10)
df.groupby("Primary Mode of Transport")[
    "Overall Satisfaction with this Commute"
].mean().sort_values(ascending=False)
df["Secondary Mode of Transport (If your trip involves multiple legs)"].value_counts().head(10)
df["Average Total Travel Time (in minutes)"].median()
```
